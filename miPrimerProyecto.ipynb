**Desarrolle** un algoritmo en Python que calcule el IMPUESTO A LAS ACTIVIDADES ECONOMICAS de acuerdo con la siguiente tabla. El monto de la actividad económica debe ser introducido por el usuario. (30%)

```python
def calcularImpuesto(monto):

    if monto >= 0.01 and monto <= 500:
        precio = 1.5
        adicional = 0
        desde = 0.01

    elif monto >= 500.01 and monto <= 1000:
        precio = 1.5
        adicional = 3
        desde = 500.01

    elif monto >= 1000.01 and monto <= 2000:
        precio = 3
        adicional = 3
        desde = 1000.01

    elif monto >= 2000.01 and monto <= 3000:
        precio = 6
        adicional = 3
        desde = 2000.01

    elif monto >= 3000.01 and monto <= 6000:
        precio = 9
        adicional = 2
        desde = 3000.01

    elif monto >= 6000.01 and monto <= 18000:
        precio = 15
        adicional = 2
        desde = 6000.01

    elif monto >= 18000.01 and monto <= 30000:
        precio = 39
        adicional = 2
        desde = 18000.01

    elif monto >= 30000.01 and monto <= 60000:
        precio = 63
        adicional = 1
        desde = 30000.01

    elif monto >= 60000.01 and monto <= 100000:
        precio = 93
        adicional = 0.8
        desde = 60000.01

    elif monto >= 100000.01 and monto <= 200000:
        precio = 125
        adicional = 0.7
        desde = 100000.01

    elif monto >= 200000.01 and monto <= 300000:
        precio = 195
        adicional = 0.6
        desde = 200000.01

    elif monto >= 300000.01 and monto <= 400000:
        precio = 255
        adicional = 0.45
        desde = 300000.01

    elif monto >= 400000.01 and monto <= 500000:
        precio = 300
        adicional = 0.4
        desde = 400000.01

    elif monto >= 500000.01 and monto <= 1000000:
        precio = 340
        adicional = 0.30
        desde = 500000.01

    elif monto >= 1000000.01 and monto <= 99999999:
        precio = 490
        adicional = 0.18
        desde = 1000000.01

    else:
        print("Monto incorrecto")
        return


    impuesto = ((monto - desde) / 1000) * adicional + precio

    print("\n------ RESULTADO ------")
    print("Monto de la actividad: $", round(monto, 2))
    print("Precio fijo: $", precio)
    print("Adicional: ", adicional)
    print("Impuesto a pagar: $", round(impuesto, 2))



monto = float(input("Ingrese el monto de la actividad economica: "))

calcularImpuesto(monto)
```

```python
from functools import reduce



# CLASE ESTADISTICA


class estadistica:


    # MEDIA ARITMETICA


    def media(self, serie):

        suma = reduce(
            lambda acum, num: acum + float(num),
            serie,
            0
        )

        return suma / len(serie)



    # MEDIANA


    def mediana(self, serie):

        serie = sorted(serie)

        cantidad = len(serie)

        if cantidad % 2 == 0:

            valor1 = float(serie[cantidad // 2 - 1])
            valor2 = float(serie[cantidad // 2])

            return (valor1 + valor2) / 2

        else:

            return float(serie[cantidad // 2])



    # MODA


    def moda(self, serie):

        mayor = 0
        moda = []

        for num in serie:

            cantidad = serie.count(num)

            if cantidad > mayor:

                mayor = cantidad
                moda = [num]

            elif cantidad == mayor and num not in moda:

                moda.append(num)

        if mayor == 1:

            return "No hay moda"

        return moda


    # VARIANZA


    def varianza(self, serie):

        media = self.media(serie)

        suma = reduce(
            lambda acum, num:
            acum + (float(num) - media) ** 2,
            serie,
            0
        )

        return suma / len(serie)



    # DESVIACION ESTANDAR


    def desviacionEstandar(self, serie):

        return self.varianza(serie) ** 0.5



    # DESVIACION TIPICA


    def desviacionTipica(self, serie):

        return self.varianza(serie) ** 0.5



    # RANGO


    def rango(self, serie):

        serie = sorted(serie)

        menor = float(serie[0])
        mayor = float(serie[-1])

        return mayor - menor



    # FRECUENCIA


    def frecuencia(self, serie):

        vistos = []

        for num in serie:

            if num not in vistos:

                cantidad = serie.count(num)

                print(
                    num,
                    "aparece",
                    cantidad,
                    "veces"
                )

                vistos.append(num)



    # MOSTRAR TODAS LAS ESTADISTICAS


    def mostrarEstadisticas(self, serie):

        print("\n------------------------------------------")

        print(
            "Media aritmetica:",
            round(self.media(serie), 2)
        )

        print(
            "Mediana:",
            round(self.mediana(serie), 2)
        )

        print(
            "Moda:",
            self.moda(serie)
        )

        print(
            "Varianza:",
            round(self.varianza(serie), 2)
        )

        print(
            "Desviacion estandar:",
            round(self.desviacionEstandar(serie), 2)
        )

        print(
            "Desviacion tipica:",
            round(self.desviacionTipica(serie), 2)
        )

        print(
            "Rango:",
            round(self.rango(serie), 2)
        )

        print("\nFrecuencia:")

        self.frecuencia(serie)

        print("------------------------------------------")



# CLASE SISTEMA ACADEMICO


class sistemaAcademico:



    # MOSTRAR TODOS LOS ALUMNOS


    def mostrarAlumnos(self, alumnos):

        for alumno in alumnos:

            print("\n==========================================")

            print("Nombre:", alumno["nombre"])
            print("Edad:", alumno["edad"])

            print("Materias:")

            for materia, nota in alumno["materias"].items():

                print(
                    " ",
                    materia,
                    ":",
                    nota
                )

            print(
                "Tiempo de traslado:",
                alumno["tiempo_traslado"],
                "minutos"
            )

            print(
                "Horas de uso del telefono:",
                alumno["horas_telefono"],
                "horas"
            )


    # ------------------------------------------------------
    # OBTENER NOTAS DE UNA MATERIA
    # ------------------------------------------------------

    def notasMateria(self, alumnos, materia):

        notas = []

        for alumno in alumnos:

            if materia in alumno["materias"]:

                notas.append(
                    alumno["materias"][materia]
                )

        return notas



    # ESTADISTICAS DE UN ESTUDIANTE


    def estadisticasEstudiante(
        self,
        alumnos,
        numero,
        objEstadistica
    ):

        alumno = alumnos[numero]

        print("\n==========================================")
        print("       DATOS DEL ESTUDIANTE")
        print("==========================================")

        print("Nombre:", alumno["nombre"])
        print("Edad:", alumno["edad"])

        print(
            "Tiempo de traslado:",
            alumno["tiempo_traslado"],
            "minutos"
        )

        print(
            "Horas de uso del telefono:",
            alumno["horas_telefono"],
            "horas"
        )

        print("\nNotas:")

        for materia, nota in alumno["materias"].items():

            print(
                materia,
                ":",
                nota
            )

        # Sacamos solamente las notas
        notas = list(
            alumno["materias"].values()
        )

        print(
            "\n========== ESTADISTICAS DE SUS NOTAS =========="
        )

        objEstadistica.mostrarEstadisticas(
            notas
        )



# DATOS DE LOS ESTUDIANTES


alumnos = [

    {
        "nombre": "Kimberly Michelle Castro Zelaya",

        "materias": {
            "Estadistica": 10,
            "Diseño de circuitos digitales": 10,
            "Ingles basico": 10,
            "Base de datos II": 10,
            "Programacion computacional III": 10
        },

        "edad": 18,
        "tiempo_traslado": 20,
        "horas_telefono": 9
    },


    {
        "nombre": "Samuel Alexander Portillo Ceron",

        "materias": {
            "Estadistica": 10,
            "Diseño de circuitos digitales": 10,
            "Ingles basico": 10,
            "Base de datos II": 10,
            "Programacion computacional III": 10
        },

        "edad": 20,
        "tiempo_traslado": 80,
        "horas_telefono": 6
    },


    {
        "nombre": "Carlos David Turcios Rivas",

        "materias": {
            "Estadistica": 9,
            "Diseño de circuitos digitales": 9,
            "Ingles basico": 10,
            "Base de datos II": 10,
            "Programacion computacional III": 10
        },

        "edad": 20,
        "tiempo_traslado": 25,
        "horas_telefono": 5
    },


    {
        "nombre": "Gilda Mileidy Gonzalez Castro",

        "materias": {
            "Estadistica": 7.5,
            "Diseño de circuitos digitales": 10,
            "Ingles basico": 10,
            "Base de datos II": 10,
            "Programacion computacional III": 10
        },

        "edad": 19,
        "tiempo_traslado": 30,
        "horas_telefono": 5.92
    },


    {
        "nombre": "Edgar Ernesto Ramirez Barrera",

        "materias": {
            "Estadistica": 10,
            "Diseño de circuitos digitales": 10,
            "Ingles basico": 10,
            "Base de datos II": 10,
            "Programacion computacional III": 10
        },

        "edad": 21,
        "tiempo_traslado": 120,
        "horas_telefono": 7
    },


    {
        "nombre": "Miguel Alejandro Gomez Saravia",

        "materias": {
            "Estadistica": 9.5,
            "Diseño de circuitos digitales": 9.5,
            "Ingles basico": 9.5,
            "Base de datos II": 9.5,
            "Programacion computacional III": 9.5
        },

        "edad": 19,
        "tiempo_traslado": 15,
        "horas_telefono": 4
    },


    {
        "nombre": "Josue Stiven Funes Benavides",

        "materias": {
            "Estadistica": 9.5,
            "Diseño de circuitos digitales": 9.5,
            "Ingles basico": 9.5,
            "Base de datos II": 9.5,
            "Programacion computacional III": 9.5
        },

        "edad": 19,
        "tiempo_traslado": 30,
        "horas_telefono": 5
    },


    {
        "nombre": "Jefferson Steven Castillo Leiva",

        "materias": {
            "Estadistica": 9.5,
            "Diseño de circuitos digitales": 9.5,
            "Ingles basico": 9.5,
            "Base de datos II": 9.5,
            "Programacion computacional III": 9.5
        },

        "edad": 19,
        "tiempo_traslado": 15,
        "horas_telefono": 2.3
    },


    {
        "nombre": "Jenifer Maricela Tenderos Guevara",

        "materias": {
            "Estadistica": 9.5,
            "Diseño de circuitos digitales": 9.5,
            "Ingles basico": 9.5,
            "Base de datos II": 9.5,
            "Programacion computacional III": 9.5
        },

        "edad": 21,
        "tiempo_traslado": 60,
        "horas_telefono": 3
    },


    {
        "nombre": "Claudia Arely Sorto Melendez",

        "materias": {
            "Estadistica": 8.5,
            "Diseño de circuitos digitales": 8.5,
            "Ingles basico": 8.5,
            "Base de datos II": 8.5,
            "Programacion computacional III": 8.5
        },

        "edad": 19,
        "tiempo_traslado": 50,
        "horas_telefono": 7.62
    },


    {
        "nombre": "Nehemias Isai Martinez Argueta",

        "materias": {
            "Estadistica": 8,
            "Diseño de circuitos digitales": 8,
            "Ingles basico": 8,
            "Base de datos II": 8,
            "Programacion computacional III": 8
        },

        "edad": 19,
        "tiempo_traslado": 45,
        "horas_telefono": 9
    },


    {
        "nombre": "Sharon Nicol Arguello Torres",

        "materias": {
            "Programacion computacional III": 8
        },

        "edad": 20,
        "tiempo_traslado": 20,
        "horas_telefono": 6
    },


    {
        "nombre": "Daniel Enrique Quintanilla Paniagua",

        "materias": {
            "Estadistica": 9.5,
            "Diseño de circuitos digitales": 9.5,
            "Ingles basico": 9.5,
            "Base de datos II": 9.5,
            "Programacion computacional III": 9.5
        },

        "edad": 19,
        "tiempo_traslado": 15,
        "horas_telefono": 4
    },


    {
        "nombre": "Jose Roberto Diaz",

        "materias": {
            "Estadistica": 9,
            "Diseño de circuitos digitales": 9,
            "Ingles basico": 9,
            "Base de datos II": 9,
            "Programacion computacional III": 9
        },

        "edad": 21,
        "tiempo_traslado": 75,
        "horas_telefono": 5
    },


    {
        "nombre": "Kevin Manrique Campos",

        "materias": {
            "Estadistica": 9.5,
            "Diseño de circuitos digitales": 9.5,
            "Ingles basico": 9.5,
            "Base de datos II": 9.5,
            "Programacion computacional III": 9.5
        },

        "edad": 19,
        "tiempo_traslado": 30,
        "horas_telefono": 6
    },


    {
        "nombre": "Jaime Chafick Salmeron",

        "materias": {
            "Programacion computacional III": 8,
            "Estadistica": 8,
            "Diseño de circuitos digitales": 10,
            "Base de datos II": 10,
            "Ingles basico": 10
        },

        "edad": 20,
        "tiempo_traslado": 7,
        "horas_telefono": 6
    },


    {
        "nombre": "Jennifer Maricela Renderos Guevara",

        "materias": {
            "Estadistica": 10,
            "Diseño de circuitos digitales": 10,
            "Ingles basico": 10,
            "Base de datos II": 10,
            "Programacion computacional III": 8
        },

        "edad": 21,
        "tiempo_traslado": 60,
        "horas_telefono": 3
    },


    {
        "nombre": "Wilfredo Josue Ortez Gaitan",

        "materias": {
            "Estadistica": 8.15,
            "Diseño de circuitos digitales": 10,
            "Ingles basico": 10,
            "Base de datos II": 10,
            "Programacion computacional III": 10
        },

        "edad": 20,
        "tiempo_traslado": 30,
        "horas_telefono": 5
    },


    {
        "nombre": "Saul Alexander Jimenez Garcia",

        "materias": {
            "Estadistica": 10,
            "Diseño de circuitos digitales": 10,
            "Ingles basico": 10,
            "Base de datos II": 10,
            "Programacion computacional III": 10
        },

        "edad": 21,
        "tiempo_traslado": 60,
        "horas_telefono": 5
    },


    {
        "nombre": "Yeimy Julissa Parada Chavez",

        "materias": {
            "Estadistica": 9.5,
            "Diseño de circuitos digitales": 9.5,
            "Ingles basico": 9.5,
            "Base de datos II": 9.5,
            "Programacion computacional III": 9.5
        },

        "edad": 20,
        "tiempo_traslado": 45,
        "horas_telefono": 3
    }

]



# CREAR OBJETOS

objEstadistica = estadistica()

objAcademico = sistemaAcademico()



# MENU PRINCIPAL


opcion = 0

while opcion != 8:

    print("\n")
    print("==========================================")
    print("       SISTEMA ESTADISTICO ACADEMICO")
    print("==========================================")

    print("1. Mostrar estudiantes")
    print("2. Estadisticas generales de edades")
    print("3. Estadisticas generales de notas")
    print("4. Estadisticas generales de traslado")
    print("5. Estadisticas generales de horas de telefono")
    print("6. Promedio por materia")
    print("7. Estadisticas de un estudiante")
    print("8. Salir")

    opcion = int(
        input("\nSeleccione una opcion: ")
    )



    # OPCION 1
    # MOSTRAR ESTUDIANTES


    if opcion == 1:

        objAcademico.mostrarAlumnos(
            alumnos
        )



    # OPCION 2
    # ESTADISTICAS DE EDADES


    elif opcion == 2:

        edades = []

        for alumno in alumnos:

            edades.append(
                alumno["edad"]
            )

        print(
            "\n========== ESTADISTICAS DE EDADES =========="
        )

        objEstadistica.mostrarEstadisticas(
            edades
        )



    # OPCION 3
    # ESTADISTICAS DE NOTAS


    elif opcion == 3:

        print(
            "\n========== MATERIAS =========="
        )

        print("1. Estadistica")
        print("2. Diseño de circuitos digitales")
        print("3. Ingles basico")
        print("4. Base de datos II")
        print("5. Programacion computacional III")

        materia = int(
            input("\nSeleccione una materia: ")
        )


        if materia == 1:

            nombreMateria = "Estadistica"


        elif materia == 2:

            nombreMateria = "Diseño de circuitos digitales"


        elif materia == 3:

            nombreMateria = "Ingles basico"


        elif materia == 4:

            nombreMateria = "Base de datos II"


        elif materia == 5:

            nombreMateria = "Programacion computacional III"


        else:

            nombreMateria = ""


        if nombreMateria != "":

            notas = objAcademico.notasMateria(
                alumnos,
                nombreMateria
            )

            print(
                "\n==========",
                nombreMateria,
                "=========="
            )

            objEstadistica.mostrarEstadisticas(
                notas
            )

        else:

            print(
                "Materia incorrecta"
            )



    # OPCION 4
    # ESTADISTICAS DE TRASLADO


    elif opcion == 4:

        traslado = []

        for alumno in alumnos:

            traslado.append(
                alumno["tiempo_traslado"]
            )

        print(
            "\n========== ESTADISTICAS DE TRASLADO =========="
        )

        objEstadistica.mostrarEstadisticas(
            traslado
        )

        print(
            "Unidad: minutos"
        )



    # OPCION 5
    # ESTADISTICAS DE TELEFONO


    elif opcion == 5:

        telefono = []

        for alumno in alumnos:

            telefono.append(
                alumno["horas_telefono"]
            )

        print(
            "\n========== ESTADISTICAS DE TELEFONO =========="
        )

        objEstadistica.mostrarEstadisticas(
            telefono
        )

        print(
            "Unidad: horas"
        )



    # OPCION 6
    # PROMEDIO POR MATERIA


    elif opcion == 6:

        print(
            "\n========== PROMEDIO POR MATERIA =========="
        )

        print("1. Estadistica")
        print("2. Diseño de circuitos digitales")
        print("3. Ingles basico")
        print("4. Base de datos II")
        print("5. Programacion computacional III")

        materia = int(
            input("\nSeleccione una materia: ")
        )


        if materia == 1:

            nombreMateria = "Estadistica"


        elif materia == 2:

            nombreMateria = "Diseño de circuitos digitales"


        elif materia == 3:

            nombreMateria = "Ingles basico"


        elif materia == 4:

            nombreMateria = "Base de datos II"


        elif materia == 5:

            nombreMateria = "Programacion computacional III"


        else:

            nombreMateria = ""


        if nombreMateria != "":

            notas = objAcademico.notasMateria(
                alumnos,
                nombreMateria
            )

            promedio = objEstadistica.media(
                notas
            )

            print(
                "\nPromedio de",
                nombreMateria,
                ":",
                round(promedio, 2)
            )

        else:

            print(
                "Materia incorrecta"
            )



    # OPCION 7
    # ESTADISTICAS DE UN ESTUDIANTE


    elif opcion == 7:

        print(
            "\n========== LISTA DE ESTUDIANTES =========="
        )

        for i in range(len(alumnos)):

            print(
                i + 1,
                ".",
                alumnos[i]["nombre"]
            )


        numero = int(
            input(
                "\nSeleccione el numero del estudiante: "
            )
        )


        if numero >= 1 and numero <= len(alumnos):

            objAcademico.estadisticasEstudiante(
                alumnos,
                numero - 1,
                objEstadistica
            )

        else:

            print(
                "Estudiante incorrecto"
            )



    # OPCION 8
    # SALIR


    elif opcion == 8:

        print(
            "\nPrograma finalizado."
        )



    # OPCION INCORRECTA


    else:

        print(
            "\nOpcion incorrecta."
        )
```
